### Comprobador de posturas ergonómicas

#### Solución menos robusta, solo recoge el angulo de piernas y cadera

```python
# Importar librerías
import cv2
import numpy as np
from ultralytics import YOLO

# 1. Cargar el modelo YOLO11 Pose
model = YOLO("yolo11m-pose.pt")

# 2. Configurar la captura de video (0 es lawebcam integrada)
cap = cv2.VideoCapture("/dev/video0", cv2.CAP_V4L2)

def calcular_angulo(p1, p2):
    x1, y1 = p1
    x2, y2 = p2
    angle = np.degrees(np.arctan2(y2 - y1, x2 - x1))
    return abs(angle)

while cap.isOpened():
    success, frame = cap.read()
    if not success: break

    results = model(frame, stream=True, conf=0.5)

    for r in results:
        keypoints = r.keypoints.xy.cpu().numpy()
        
        for person_kpts in keypoints:
            # Verificamos que existan puntos (evitar ruidos de detección)
            if len(person_kpts) < 13: continue 
            
            # Puntos: 5=Hombro Izq, 11=Cadera Izq, 0=Nariz (para posicionar texto)
            hombro = person_kpts[5]
            cadera = person_kpts[11]
            nariz = person_kpts[0] # Usaremos la nariz para el texto

            if hombro[1] > 0 and cadera[1] > 0:
                angulo_torso = calcular_angulo(hombro, cadera)
                
                # Lógica de color
                if 70 < angulo_torso < 100:
                    color = (0, 255, 0)
                    status = "OK"
                else:
                    color = (0, 0, 255)
                    status = "Mala Postura"

                # Coordenadas para el texto: 
                # Si la nariz no se detecta, usamos el hombro pero un poco más arriba
                text_x = int(hombro[0])
                text_y = int(hombro[1] - 20) if nariz[1] == 0 else int(nariz[1] - 40)

                # Dibujar línea en el torso
                cv2.line(frame, (int(hombro[0]), int(hombro[1])), 
                         (int(cadera[0]), int(cadera[1])), color, 3)
                
                # Dibujar el texto flotando sobre la persona
                cv2.putText(frame, f"{status} ({int(angulo_torso)} deg)", 
                            (text_x, text_y), 
                            cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)

    cv2.imshow("Monitor Multi-Persona", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'): break

cap.release()
cv2.destroyAllWindows()
```
