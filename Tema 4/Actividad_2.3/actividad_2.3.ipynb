### Pasando de texto a voz

```python
# Instalando librerias
%pip install edge-tts
```

```python
import asyncio
import edge_tts

# Seleccionando modelo
voz = "es-ES-AlvaroNeural"   # también puedes usar es-ES-ElviraNeural

# Cargando texto desde un archivo
with open("texto_prueba.txt", "r", encoding="utf-8") as f:
    texto = f.read()

async def main():
    communicate = edge_tts.Communicate(texto, voz)
    await communicate.save("salida.mp3")

await main()
print("Audio guardado en salida.mp3")
```
