#### Impacto del Rol System (Groq)

```python
import os
from groq import Groq
from dotenv import load_dotenv

load_dotenv()
client = Groq()

codigo_a_revisar = "def suma(a, b): return a+b"

# CONFIGURACIÓN A: Sin rol de sistema (comportamiento genérico)
res_generica = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
    {"role": "user", "content": f"Revisa este código: {codigo_a_revisar}"}
 ]
)

# CONFIGURACIÓN B: Con rol de sistema (comportamiento experto)
res_experta = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
    {
        "role": "system",
        "content": "Eres un Ingeniero de Software Senior en Google. Tu tono es crítico, técnico y breve. Solo respondes con sugerencias de optimización o errores."
    },
    {"role": "user", "content": f"Revisa este código:{codigo_a_revisar}"}]
)
print("--- RESPUESTA GENÉRICA ---")
print(res_generica.choices[0].message.content)
print("\n--- RESPUESTA EXPERTA (SYSTEM ROLE) ---")
print(res_experta.choices[0].message.content)
```

#### Memoria y el rol Assistant (Groq)

```python
# Continuando en el mismo archivo o uno nuevo
historial = [   
    {"role": "system", "content": "Eres un tutor de Python."},
    {"role": "user", "content": "Hola, me llamo Carlos y quiero aprender listas."},
    {"role": "assistant", "content": "¡Hola Carlos! Encantado. Una lista es una colección ordenada. ¿Quieres ver un ejemplo?"}, 
    {"role": "user", "content": "¿Cómo me llamo y qué estábamos hablando?"}
]

response = client.chat.completions.create(
 model="llama-3.3-70b-versatile",
 messages=historial
)

print("\n--- PRUEBA DE MEMORIA ---")
print(response.choices[0].message.content)
```

#### Realizando razonamiento mas avanzado con API Key de Gemini

```python
from google import genai
import os

# Asegúrate de que la clave se cargue correctamente desde el entorno
api_key = os.getenv('GEMINI_API_KEY')

# Ahora crea el cliente explicitamente pasando la API key
client = genai.Client(api_key=api_key)

# Ejemplo de uso con Gemini 2.5 Flash
prompt_tecnico = """
Explica el concepto de 'Decoradores' en Python.
Incluye un ejemplo de código y cómo se aplicaría en un entorno real
(ej. logging).
Responde en formato Markdown.
"""

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt_tecnico
)

print("--- ANÁLISIS DE GEMINI 2.5 ---")
print(response.text)
```

#### Tarea Final: El Comparador Multimodal

```python
from google import genai
from dotenv import load_dotenv
import os

# Cargar las variables de entorno (como la clave API de Gemini)
load_dotenv()

# Cliente para Gemini (ya configurado correctamente)
gemini_client = genai.Client(api_key=os.getenv('GEMINI_API_KEY'))

# 1. Usando el cliente de Groq para realizar una solicitud
prompt_tecnico = """
Explica el concepto de arrays en Python y define sus principales usos.
Incluye un ejemplo de código y cómo se aplicaría en un entorno real.
Responde en formato Markdown.
"""

# Cliente de Groq (específico para la interacción con Groq)
groq_response = groq_client.chat.completions.create(
    model="openai/gpt-oss-120b",  # Este es el modelo de Groq
    messages=[
        {"role": "user", "content": prompt_tecnico}
    ],
    temperature=0.7
)

print("\n--- RESPUESTA DE GROQ ---")
print(groq_response.choices[0].message.content)

# 2. Usando Groq con un tono sarcástico
res_sarcastica = groq_client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
    {
        "role": "system",
        "content": "Eres un Ingeniero de Software Senior. Tu tono es sarcástico, conciso y breve."
    },
    {"role": "user", "content": f"Revisa este código:{prompt_tecnico}"}
])

print("\n--- RESPUESTA SARCASTICA (SYSTEM ROLE) ---")
print(res_sarcastica.choices[0].message.content)

# 3. Usando Gemini (modelo gemini-2.5-pro) para una explicación académica
# Usamos `generate_content()` en lugar de `chat.completions.create()`
gemini_response = gemini_client.models.generate_content(
    model="gemini-2.5-flash",  # Este es el modelo de Gemini
    contents=prompt_tecnico  # Mensaje para generar el contenido
)

print("\n--- RESPUESTA ACADÉMICA DE GEMINI (Modelo Gemini 2.5 Flash) ---")
print(gemini_response.text)
```
